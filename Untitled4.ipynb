```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# 1. Generate a clean, realistic Student Performance Dataset locally
# This avoids any external URL/network download errors in Colab!
np.random.seed(42)
n_samples = 600

data = {
    'studytime': np.random.choice([1, 2, 3, 4], size=n_samples, p=[0.2, 0.4, 0.3, 0.1]),
    'failures': np.random.choice([0, 1, 2, 3], size=n_samples, p=[0.7, 0.2, 0.05, 0.05]),
    'absences': np.random.poisson(lam=4, size=n_samples),
    'health': np.random.choice([1, 2, 3, 4, 5], size=n_samples),
    'internet': np.random.choice(['yes', 'no'], size=n_samples, p=[0.8, 0.2]),
    'higher': np.random.choice(['yes', 'no'], size=n_samples, p=[0.9, 0.1]),
    'romantic': np.random.choice(['yes', 'no'], size=n_samples, p=[0.3, 0.7])
}

df = pd.DataFrame(data)

# Create a realistic target variable 'at_risk' (1 = at risk of failing, 0 = passing)
# Factors like high absences and low study time increase risk
risk_score = (df['absences'] * 0.3) + (df['failures'] * 1.5) - (df['studytime'] * 0.8) + np.random.normal(0, 1, n_samples)
df['at_risk'] = (risk_score > risk_score.median()).astype(int)

print(f"Dataset successfully created! Shape: {df.shape}")
display(df.head(3))

# 2. Handle Missing Values (Safety check)
df = df.dropna()

# 3. Separate Features (X) and Target (y)
X = df.drop(columns=['at_risk'])
y = df['at_risk']

# 4. Encode Categorical Variables
categorical_cols = X.select_dtypes(include=['object']).columns
X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

print(f"\nFeatures shape after encoding: {X_encoded.shape}")

# 5. Split the Data into Training and Testing Sets (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

print(f"\nSetup complete and working perfectly!")
print(f"Training set: {X_train.shape}")
print(f"Testing set: {X_test.shape}")
print(f"At-risk proportion: {y_train.mean():.2%}")
```

```python
# Import necessary libraries for model training and evaluation
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score, confusion_matrix

# 1. Initialize the Random Forest Classifier
# We set random_state for reproducibility
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=6,
    random_state=42
)

# 2. Train (Fit) the model on the training data
print("Training the Random Forest model...")
rf_model.fit(X_train, y_train)
print("Model training complete!")

# 3. Make predictions on the testing set
y_pred = rf_model.predict(X_test)
y_prob = rf_model.predict_proba(X_test)[:, 1] # Probability of being 'at_risk'

# 4. Evaluate Baseline Performance
accuracy = accuracy_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_prob)

print(f"\n--- Baseline Model Evaluation ---")
print(f"Accuracy: {accuracy:.4f}")
print(f"ROC-AUC Score: {roc_auc:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# Import necessary libraries
import shap
import matplotlib.pyplot as plt

# 1. Initialize the SHAP TreeExplainer for our trained Random Forest model
print("Initializing SHAP TreeExplainer...")
explainer = shap.TreeExplainer(rf_model)

# Calculate SHAP values for the test set
shap_values = explainer(X_test)

print("SHAP values calculated successfully!")

# 2. Generate Global Feature Importance Plot (Summary Bar Plot)
print("\nGenerating Global Feature Importance Plot...")
plt.figure(figsize=(10, 6))
shap.plots.bar(shap_values[:, :, 1] if len(shap_values.shape) == 3 else shap_values, show=False)
plt.title("Global Feature Importance (SHAP)", fontsize=14)
plt.tight_layout()
plt.show()

# 3. Generate SHAP Beeswarm Plot (Shows direction and magnitude)
print("\nGenerating SHAP Beeswarm Plot...")
plt.figure(figsize=(10, 6))
shap.plots.beeswarm(shap_values[:, :, 1] if len(shap_values.shape) == 3 else shap_values, show=False)
plt.title("SHAP Feature Impact Distribution", fontsize=14)
plt.tight_layout()
plt.show()

# 4. Generate Individual Student Prediction Breakdown (Waterfal/Force Plot)
# We look at the first student in the test set using the modern shap.plots interface
print(f"\nGenerating Individual Explanation for Student Index 0:")
student_index = 0

plt.figure(figsize=(10, 4))
# Using waterfall plot which is the modern, stable replacement for force_plot in recent SHAP versions
single_explanation = shap_values[student_index, :, 1] if len(shap_values.shape) == 3 else shap_values[student_index]
shap.plots.waterfall(single_explanation, show=True)
```

```python
# 1. Install Gradio if not already installed
!pip install gradio -q

import gradio as gr
import pandas as pd
import numpy as np
import shap
import matplotlib.pyplot as plt

# Define the prediction and explanation function for Gradio
def predict_student_risk(studytime, failures, absences, health, internet, higher, romantic):
    # Create a DataFrame from user inputs
    input_data = pd.DataFrame([{
        'studytime': int(studytime),
        'failures': int(failures),
        'absences': int(absences),
        'health': int(health),
        'internet_yes': 1 if internet == 'yes' else 0,
        'higher_yes': 1 if higher == 'yes' else 0,
        'romantic_yes': 1 if romantic == 'yes' else 0
    }])

    # Align columns with model training data structure
    input_data = input_data.reindex(columns=X_train.columns, fill_value=0)

    # Predict probability and class
    prob = rf_model.predict_proba(input_data)[0][1]
    pred = rf_model.predict(input_data)[0]

    # Text output status
    status = "🚨 AT-RISK (Needs Intervention)" if pred == 1 else "✅ SAFE (Passing Track)"
    risk_text = f"**Prediction:** {status}\n\n**Failure Probability:** {prob * 100:.1f}%"

    # Generate SHAP Waterfall plot for this specific input
    explainer = shap.TreeExplainer(rf_model)
    shap_vals = explainer(input_data)

    plt.figure(figsize=(9, 4))
    single_val = shap_vals[0, :, 1] if len(shap_vals.shape) == 3 else shap_vals[0]
    shap.plots.waterfall(single_val, show=False)
    plt.title("Why did the model make this decision? (SHAP Breakdown)", fontsize=11)
    plt.tight_layout()

    fig = plt.gcf()
    return risk_text, fig

# 2. Build the Gradio Interface Components
app = gr.Interface(
    fn=predict_student_risk,
    inputs=[
        gr.Slider(1, 4, step=1, value=2, label="Study Time (1: <2hrs, 2: 2-5hrs, 3: 5-10hrs, 4: >10hrs)"),
        gr.Slider(0, 3, step=1, value=0, label="Past Class Failures"),
        gr.Slider(0, 30, step=1, value=4, label="Number of Absences"),
        gr.Slider(1, 5, step=1, value=3, label="Current Health Status (1-5)"),
        gr.Radio(["yes", "no"], value="yes", label="Internet Access at Home"),
        gr.Radio(["yes", "no"], value="yes", label="Wants Higher Education"),
        gr.Radio(["yes", "no"], value="no", label="In a Romantic Relationship")
    ],
    outputs=[
        gr.Markdown(label="Risk Assessment"),
        gr.Plot(label="Explainable AI (SHAP) Waterfall Explanation")
    ],
    title="🎓 Student Performance Risk Predictor & XAI Analyzer",
    description="Adjust student parameters below using sliders and radio buttons. The model will instantly predict their academic risk level and generate a SHAP breakdown explaining the exact factors driving the prediction."
)

# 3. Launch the App (Generates a local link and a shareable public live link!)
app.launch(inline=False, share=True)
```
